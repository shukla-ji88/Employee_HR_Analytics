```python

# 1. Dataset ki pehli 5 rows
print("FIRST 5 EMPLOYEES")
display(df.head())

# 2. Total rows aur columns
print("\nDATASET SIZE:", df.shape)

# 3. Column names aur data types
print("\nDATA TYPES")
df.info()

# 4. Missing values check
print("\nMISSING VALUES")
print(df.isnull().sum())

# 5. Duplicate records check
print("\nDUPLICATE ROWS:", df.duplicated().sum())

# 6. Salary aur experience ka statistical summary
print("\nSTATISTICAL SUMMARY")
display(df[["Salary", "ExperienceYears", "PerformanceRating"]].describe())
```

```python

import matplotlib.pyplot as plt

# 1. Department-wise analysis
dept_analysis = df.groupby("Department").agg(
    Total_Employees=("EmployeeID", "count"),
    Average_Salary=("Salary", "mean"),
    Total_Payroll=("Salary", "sum")
).round(2)

print("DEPARTMENT-WISE ANALYSIS")
display(dept_analysis)

# 2. Employees by department - Bar Chart
dept_analysis["Total_Employees"].sort_values(
    ascending=False
).plot(kind="bar", figsize=(8, 4), title="Employees by Department")

plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 3. Average salary by department - Bar Chart
dept_analysis["Average_Salary"].sort_values(
    ascending=False
).plot(kind="bar", figsize=(8, 4), title="Average Salary by Department")

plt.xlabel("Department")
plt.ylabel("Average Salary (INR)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

import matplotlib.pyplot as plt

# 1. Top 5 highest-paid employees
print("TOP 5 HIGHEST-PAID EMPLOYEES")
display(
    df.nlargest(5, "Salary")[
        ["Name", "Department", "Salary", "ExperienceYears"]
    ]
)

# 2. Lowest 5 salaries
print("TOP 5 LOWEST-PAID EMPLOYEES")
display(
    df.nsmallest(5, "Salary")[
        ["Name", "Department", "Salary", "ExperienceYears"]
    ]
)

# 3. Experience aur salary ke beech relationship
correlation = df["ExperienceYears"].corr(df["Salary"])
print("Experience-Salary Correlation:", round(correlation, 2))

# 4. Performance rating ke hisaab se salary
performance_analysis = df.groupby("PerformanceRating").agg(
    Employee_Count=("EmployeeID", "count"),
    Average_Salary=("Salary", "mean")
).round(2)

print("PERFORMANCE-WISE SALARY ANALYSIS")
display(performance_analysis)

# 5. Experience vs Salary scatter plot
plt.figure(figsize=(8, 5))
plt.scatter(df["ExperienceYears"], df["Salary"])
plt.title("Employee Experience vs Salary")
plt.xlabel("Experience (Years)")
plt.ylabel("Salary (INR)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# 6. Average salary by performance rating
performance_analysis["Average_Salary"].plot(
    kind="bar",
    figsize=(8, 4),
    title="Average Salary by Performance Rating"
)
plt.xlabel("Performance Rating")
plt.ylabel("Average Salary (INR)")
plt.tight_layout()
plt.show()
```

```python

# STEP 7: BUSINESS ANALYSIS

# Query 1: Employees with salary above 50,000
print("1. EMPLOYEES WITH SALARY ABOVE 50000")

high_salary = df.query("Salary > 50000").sort_values(
    by="Salary", ascending=False
)
display(high_salary[["Name", "Department", "Salary"]])


# Query 2: IT department employees
print("2. IT DEPARTMENT EMPLOYEES")

it_employees = df.query("Department == 'IT'")
display(it_employees[["Name", "Position", "Salary"]])


# Query 3: Department-wise salary ranking
print("3. DEPARTMENT-WISE SALARY RANKING")

dept_salary = (
    df.groupby("Department")["Salary"]
    .mean()
    .sort_values(ascending=False)
    .reset_index(name="Average_Salary")
)
display(dept_salary)


# Query 4: High-performing employees
print("4. EMPLOYEES WITH PERFORMANCE RATING 5")

top_performers = df.query("PerformanceRating == 5")
display(top_performers[["Name", "Department", "Salary"]])


# Query 5: Overall business insights
total_employees = df["EmployeeID"].nunique()
average_salary = df["Salary"].mean()
total_payroll = df["Salary"].sum()
highest_paid_employee = df.loc[df["Salary"].idxmax()]
largest_department = df["Department"].value_counts().idxmax()
highest_avg_salary_dept = dept_salary.iloc[0]["Department"]

print("\nBUSINESS INSIGHTS")
print("Total Employees:", total_employees)
print("Average Salary:", round(average_salary, 2))
print("Total Salary Amount:", total_payroll)
print("Largest Department:", largest_department)
print("Highest Average Salary Department:", highest_avg_salary_dept)
print(
    "Highest Paid Employee:",
    highest_paid_employee["Name"],
    "- Salary:",
    highest_paid_employee["Salary"]
)


# Query 6: Export results for reporting
dept_salary.to_csv("department_salary_report.csv", index=False)
high_salary.to_csv("high_salary_employees.csv", index=False)

print("\nReports exported successfully!")
```
